# 02 — Exploratory spatial and temporal analysis

Explore the simulated county-month panel before modeling. All examples here use synthetic environmental values and outcomes. Point locations are illustrative and do not represent true county geography. A real-data extension would need validated county boundaries and joins; `viz.choropleth_counties` provides a plotting helper.

```python
import sys
from pathlib import Path
sys.path.append(str(Path.cwd().parent))

import pandas as pd
from src import config
from src import viz

from src.features import load_panel
panel = load_panel()
```

## 1. Study area and landscape context (agricultural share)

```python
agg = panel.groupby("county").agg(lat=("lat","first"), lon=("lon","first"),
                                    ag_prop=("ag_prop","first"), detections=("y","sum")).reset_index()
viz.map_panel(agg, "ag_prop", "Agricultural land share by county", config.FIGURES/"fig1_study_area.png",
              cmap="YlGn", label="Ag share")
```

## 2. Wildfire exposure (mean monthly fire detections by county)

```python
fire_agg = panel.groupby("county").agg(lat=("lat","first"), lon=("lon","first"),
                                        fire=("fire_count","mean")).reset_index()
viz.map_panel(fire_agg, "fire", "Mean monthly active-fire detections", config.FIGURES/"fig2_fire_exposure.png",
              cmap="YlOrRd", label="Fires / month")
```

## 3. Temporal co-occurrence: detections vs. fire season

```python
viz.temporal_trend(panel, config.FIGURES/"fig3_temporal_trends.png")
```

**Interpretation:** these figures illustrate exploratory checks on synthetic data. They do not establish observed seasonality or an environmental association in California. A real-data extension should assess lag choices and surveillance coverage before interpreting patterns.
